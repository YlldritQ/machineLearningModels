```python
# Imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import (
    classification_report, confusion_matrix,
    accuracy_score, roc_auc_score, roc_curve
)
from IPython.display import display, clear_output
import ipywidgets as widgets

# Load and preprocess data
columns = ['pregnancies', 'glucose', 'blood_pressure', 'skin_thickness',
           'insulin', 'bmi', 'diabetes_pedigree', 'age', 'outcome']
data = pd.read_csv('pima-indians-diabetes.data', delimiter=',', names=columns, header=None)

# Clean invalid zeros
invalid_zero_cols = ['glucose', 'blood_pressure', 'skin_thickness', 'insulin', 'bmi']
data[invalid_zero_cols] = data[invalid_zero_cols].replace(0, np.nan)
data.fillna(data.median(numeric_only=True), inplace=True)

# Split and scale data
X = data.drop('outcome', axis=1)
y = data['outcome']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# MLP Model
model = MLPClassifier(hidden_layer_sizes=(32, 16, 8), activation='relu',
                      solver='adam', max_iter=1, warm_start=True,
                      early_stopping=False, random_state=42)

# Manual training loop with visualization
epochs = 300
train_acc, test_acc, train_loss = [], [], []
classes = np.unique(y_train)

# Widgets
progress = widgets.IntProgress(value=0, min=0, max=epochs, description='Training')
status = widgets.Label(value="Starting training...")
display(progress, status)

plt.ion()
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

for epoch in range(epochs):
    model.partial_fit(X_train_scaled, y_train, classes=classes)
    y_train_pred = model.predict(X_train_scaled)
    y_test_pred = model.predict(X_test_scaled)

    train_acc.append(accuracy_score(y_train, y_train_pred))
    test_acc.append(accuracy_score(y_test, y_test_pred))
    train_loss.append(model.loss_)

    # Update progress bar and label
    progress.value = epoch + 1
    status.value = f"Epoch {epoch+1}/{epochs} - Train Acc: {train_acc[-1]:.4f} - Test Acc: {test_acc[-1]:.4f} - Loss: {train_loss[-1]:.4f}"

    # Real-time plot update
    clear_output(wait=True)
    display(progress, status)

    ax1.clear()
    ax2.clear()

    ax1.plot(train_acc, label="Train Accuracy")
    ax1.plot(test_acc, label="Test Accuracy")
    ax1.set_title("Accuracy Over Epochs")
    ax1.set_xlabel("Epoch")
    ax1.set_ylabel("Accuracy")
    ax1.legend()

    ax2.plot(train_loss, label="Training Loss", color='orange')
    ax2.set_title("Training Loss Over Epochs")
    ax2.set_xlabel("Epoch")
    ax2.set_ylabel("Loss")

    plt.tight_layout()
    plt.pause(0.001)

plt.ioff()
plt.show()

# Final predictions
y_pred = model.predict(X_test_scaled)
y_prob = model.predict_proba(X_test_scaled)[:, 1]

print("\nClassification Report:\n", classification_report(y_test, y_pred))

# Confusion Matrix
plt.figure(figsize=(6, 5))
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.show()

# ROC Curve
fpr, tpr, _ = roc_curve(y_test, y_prob)
auc_score = roc_auc_score(y_test, y_prob)

plt.figure(figsize=(7, 5))
plt.plot(fpr, tpr, label=f"ROC AUC = {auc_score:.2f}")
plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
plt.title("ROC Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.grid(True)
plt.show()

# Predict examples
example_columns = ['pregnancies', 'glucose', 'blood_pressure', 'skin_thickness',
                   'insulin', 'bmi', 'diabetes_pedigree', 'age']

high_risk = pd.DataFrame([[10, 180, 90, 35, 200, 40.0, 1.5, 60]], columns=example_columns)
low_risk = pd.DataFrame([[1, 85, 60, 20, 80, 22.0, 0.2, 25]], columns=example_columns)

high_risk_scaled = scaler.transform(high_risk)
low_risk_scaled = scaler.transform(low_risk)

print("\nExample Predictions:")
for i, (person_scaled, label_raw) in enumerate([(high_risk_scaled, "High Risk"), (low_risk_scaled, "Low Risk")], 1):
    pred = model.predict(person_scaled)[0]
    prob = model.predict_proba(person_scaled)[0][1]
    label = "Diabetic" if pred == 1 else "Non-Diabetic"
    print(f"{label_raw} Person: Prediction = {label} (Probability = {prob:.2%})")

# Show predictions for the last 5 entries in the test set
print("\nLast 5 Samples from Test Set with Predictions:")
last_5 = X_test.tail(5)
last_5_scaled = scaler.transform(last_5)

preds = model.predict(last_5_scaled)
probs = model.predict_proba(last_5_scaled)[:, 1]

for i, (input_vals, pred_label, prob_val) in enumerate(zip(last_5.values, preds, probs), 1):
    label = "Diabetic" if pred_label == 1 else "Non-Diabetic"
    print(f"Sample {i}: Features = {input_vals} -> Prediction = {label} (Probability = {prob_val:.2%})")
```

```python
def get_user_input_and_predict(model, scaler):
    print("\n🔢 Enter patient information to predict diabetes:")
    try:
        user_data = {
            'pregnancies': int(input("Pregnancies: ")),
            'glucose': float(input("Glucose level (mg/dL): ")),
            'blood_pressure': float(input("Blood Pressure (mm Hg): ")),
            'skin_thickness': float(input("Skin Thickness (mm): ")),
            'insulin': float(input("Insulin (mu U/ml): ")),
            'bmi': float(input("BMI (kg/m²): ")),
            'diabetes_pedigree': float(input("Diabetes Pedigree Function: ")),
            'age': int(input("Age (years): "))
        }

        # Convert to DataFrame and scale
        input_df = pd.DataFrame([user_data])
        input_scaled = scaler.transform(input_df)

        # Predict
        pred = model.predict(input_scaled)[0]
        prob = model.predict_proba(input_scaled)[0][1]
        label = "Diabetic" if pred == 1 else "Non-Diabetic"

        print(f"\n📊 Prediction: {label} (Probability = {prob:.2%})")

    except ValueError:
        print("⚠️ Invalid input. Please enter numbers only.")

# Call this after training and evaluation
get_user_input_and_predict(model, scaler)
```

```python
from sklearn.linear_model import Perceptron
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

# Helper function to evaluate classifiers
def evaluate_classifier(model, name):
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_test_scaled)
    y_prob = None

    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test_scaled)[:, 1]
    elif hasattr(model, "decision_function"):
        # Perceptron has decision_function instead of predict_proba
        scores = model.decision_function(X_test_scaled)
        y_prob = (scores - scores.min()) / (scores.max() - scores.min() + 1e-8)  # Min-max normalization

    print(f"\n{name} - Classification Report:\n")
    print(classification_report(y_test, y_pred))

    # Confusion Matrix
    plt.figure(figsize=(6, 5))
    sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Greens')
    plt.title(f"{name} - Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.show()

    # ROC Curve
    if y_prob is not None:
        fpr, tpr, _ = roc_curve(y_test, y_prob)
        auc_score = roc_auc_score(y_test, y_prob)

        plt.figure(figsize=(7, 5))
        plt.plot(fpr, tpr, label=f"{name} ROC AUC = {auc_score:.2f}")
        plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
        plt.title(f"{name} - ROC Curve")
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.legend()
        plt.grid(True)
        plt.show()
    else:
        print(f"{name} does not support probability prediction; ROC AUC not available.")

# Instantiate and evaluate classifiers
perceptron = Perceptron(max_iter=1000, random_state=42)
knn = KNeighborsClassifier(n_neighbors=5)
decision_tree = DecisionTreeClassifier(random_state=42)

# Evaluate all models
evaluate_classifier(perceptron, "Perceptron")
evaluate_classifier(knn, "k-Nearest Neighbors (k=5)")
evaluate_classifier(decision_tree, "Decision Tree")
```

```python
from sklearn.model_selection import LeaveOneOut
from tqdm import tqdm

# LOOCV Function
def leave_one_out_cv(model, X, y, name):
    loo = LeaveOneOut()
    correct = 0
    probs = []
    true_labels = []
    supports_proba = hasattr(model, "predict_proba") or hasattr(model, "decision_function")

    print(f"\nRunning Leave-One-Out Cross-Validation for {name}...")
    for train_index, test_index in tqdm(loo.split(X), total=len(X)):
        X_train_loo, X_test_loo = X[train_index], X[test_index]
        y_train_loo, y_test_loo = y[train_index], y[test_index]

        model.fit(X_train_loo, y_train_loo)
        y_pred = model.predict(X_test_loo)

        correct += int(y_pred[0] == y_test_loo[0])
        true_labels.append(y_test_loo[0])

        if supports_proba:
            if hasattr(model, "predict_proba"):
                prob = model.predict_proba(X_test_loo)[0][1]
            else:
                score = model.decision_function(X_test_loo)[0]
                # Normalize to 0-1
                prob = (score - score) / 1e-8  # Since only one sample, this ends up as 0
            probs.append(prob)

    accuracy = correct / len(y)
    print(f"\n{name} - LOOCV Accuracy: {accuracy:.4f}")

    # ROC Curve if probabilities available
    if supports_proba and len(set(true_labels)) > 1:
        auc_score = roc_auc_score(true_labels, probs)
        fpr, tpr, _ = roc_curve(true_labels, probs)

        plt.figure(figsize=(7, 5))
        plt.plot(fpr, tpr, label=f"{name} ROC AUC = {auc_score:.2f}")
        plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
        plt.title(f"{name} - LOOCV ROC Curve")
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.legend()
        plt.grid(True)
        plt.show()

# Apply LOOCV on each model
leave_one_out_cv(perceptron, X_train_scaled, y_train.values, "Perceptron")
leave_one_out_cv(knn, X_train_scaled, y_train.values, "k-Nearest Neighbors (k=5)")
leave_one_out_cv(decision_tree, X_train_scaled, y_train.values, "Decision Tree")
```

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import confusion_matrix, adjusted_rand_score
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv('pima-indians-diabetes.data', delimiter=',', names=columns, header=None)

# Drop ID and Outcome
X = data.drop(['outcome'], axis=1)
y = data['outcome']
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

```python
for k in range(2, 7):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    clusters = kmeans.fit_predict(X_scaled)

    plt.figure(figsize=(6, 4))
    plt.scatter(X_pca[:, 0], X_pca[:, 1], c=clusters, cmap='Set1', s=50)
    plt.title(f'K-Means Clustering (k={k})')
    plt.xlabel('PCA 1')
    plt.ylabel('PCA 2')
    plt.grid(True)
    plt.show()
```

```python
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_scaled)

# Visualize final clusters
plt.figure(figsize=(6, 4))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=clusters, cmap='Set1', s=50)
plt.title('Final K-Means Clustering (k=2)')
plt.xlabel('PCA 1')
plt.ylabel('PCA 2')
plt.grid(True)
plt.show()
```

```python
# Confusion matrix
cm = confusion_matrix(y, clusters)
print("Confusion Matrix (clustering vs true labels):\n", cm)

# Adjusted Rand Index
ari = adjusted_rand_score(y, clusters)
print(f"Adjusted Rand Index (k=2): {ari:.4f}")
```

```python
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.model_selection import LeaveOneOut
from tqdm import tqdm
import numpy as np
import warnings
from sklearn.exceptions import ConvergenceWarning

warnings.filterwarnings("ignore", category=ConvergenceWarning)

def model_performance_summary(models, X_train, y_train, X_test, y_test):
    results = []

    def loocv_accuracy(model, X, y):
        loo = LeaveOneOut()
        correct = 0
        for train_idx, test_idx in loo.split(X):
            model.fit(X[train_idx], y[train_idx])
            pred = model.predict(X[test_idx])
            correct += int(pred[0] == y[test_idx][0])
        return correct / len(y)

    for name, model in models.items():
        # Train on training data
        model.fit(X_train, y_train)

        # Predict on test data
        y_pred = model.predict(X_test)
        acc = accuracy_score(y_test, y_pred)

        # Get probability or decision scores for ROC AUC if possible
        if hasattr(model, "predict_proba"):
            y_prob = model.predict_proba(X_test)[:, 1]
            auc = roc_auc_score(y_test, y_prob)
        elif hasattr(model, "decision_function"):
            scores = model.decision_function(X_test)
            # Min-max normalize decision scores to [0,1]
            y_prob = (scores - scores.min()) / (scores.max() - scores.min() + 1e-8)
            auc = roc_auc_score(y_test, y_prob)
        else:
            auc = None

        # Calculate LOOCV accuracy on training set (this can be slow)
        loo_acc = loocv_accuracy(model, X_train, y_train)

        results.append({
            "Model": name,
            "Test Accuracy": acc,
            "Test ROC AUC": auc,
            "LOOCV Accuracy": loo_acc
        })

    # Print summary table
    print(f"{'Model':<30} {'Test Acc':<10} {'Test ROC AUC':<15} {'LOOCV Acc':<10}")
    print("-" * 70)
    for r in results:
        auc_str = f"{r['Test ROC AUC']:.4f}" if r['Test ROC AUC'] is not None else "N/A"
        print(f"{r['Model']:<30} {r['Test Accuracy']:<10.4f} {auc_str:<15} {r['LOOCV Accuracy']:<10.4f}")

# Define your models
models = {
    "MLP Classifier": model,
    "Perceptron": perceptron,
    "k-Nearest Neighbors (k=5)": knn,
    "Decision Tree": decision_tree
}

# Run the comparison summary
model_performance_summary(models, X_train_scaled, y_train.values, X_test_scaled, y_test.values)
```
